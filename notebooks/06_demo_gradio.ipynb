Launches the DocRescue interactive demo with a public shareable link via Gradio.

**Pre-requisite**: `experiments/checkpoints/best.pth` must be in your Drive.

````python
!pip -q install gradio segmentation-models-pytorch albumentations opencv-python pytesseract easyocr pyyaml
import subprocess, sys
if 'google.colab' in sys.modules:
    subprocess.run(['apt-get', '-qq', 'install', 'tesseract-ocr'], check=True)
````

````python
import os, sys, yaml

COLAB = 'google.colab' in sys.modules
if COLAB:
    from google.colab import drive
    drive.mount('/content/drive')
    PROJECT_ROOT = '/content/drive/MyDrive/DocRescue'
else:
    PROJECT_ROOT = '.'

sys.path.insert(0, PROJECT_ROOT)

with open(os.path.join(PROJECT_ROOT, 'configs/config.yaml')) as f:
    cfg = yaml.safe_load(f)

CKPT = os.path.join(PROJECT_ROOT, 'experiments/checkpoints/best.pth')
print(f'Checkpoint: {CKPT}, exists={os.path.exists(CKPT)}')
````

````python
# ── Launch demo ─────────────────────────────────────────────────────────────
import sys
sys.path.insert(0, PROJECT_ROOT)

# Import the build_ui function from app/app.py
import importlib.util
spec = importlib.util.spec_from_file_location(
    'app', os.path.join(PROJECT_ROOT, 'app', 'app.py')
)
app_module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(app_module)

demo = app_module.build_ui(CKPT, cfg)

# share=True gives a public URL valid for 72 hours
demo.launch(share=True, inline=True)
````
